**4.1** Consider the optimization problem
$$
\begin{array}{ll}
\text{minimize} & f_0(x_1, x_2) \\
\text{subject to} & 2x_1 + x_2 \geq 1 \\
& x_1 + 3x_2 \geq 1 \\
& x_1 \geq 0, \quad x_2 \geq 0.
\end{array}
$$

Make a sketch of the feasible set. For each of the following objective functions, give the optimal set and the optimal value.



```python
import numpy as np
import cvxpy as cp
import matplotlib.pyplot as plt
```

```python
x1 = cp.Variable(1)
x2 = cp.Variable(1)

const = [
    2*x1 + x2 >= 1,
    x1 + 3*x2 >= 1,
    x1 >= 0,
    x2 >= 0
]
```


(a) $f_0(x_1, x_2) = x_1 + x_2$.

```python
f0a = x1 + x2

cost = cp.Minimize(f0a)

prob = cp.Problem(cost, const)
prob.solve()

print("Solver Status:", prob.status)
if prob.status != 'optimal':
    print("⚠️ Optimization did not converge. Results are invalid.")
else:
    ax1 = x1.value
    ax2 = x2.value
    minf0a = ax1 + ax2    
    np.set_printoptions(precision=6, suppress=False)
    print("Best solution x1* = ", ax1)
    print("Best solution x2* = ", ax2)
    print("Best objective f0* = ",minf0a)

```


(b) $f_0(x_1, x_2) = -x_1 - x_2$.


```python
f0b = (-1)*x1 - x2

cost = cp.Minimize(f0b)

prob = cp.Problem(cost, const)
prob.solve()

print("Solver Status:", prob.status)
if prob.status != 'optimal':
    print("⚠️ Optimization did not converge. Results are invalid.")
else:
    bx1 = x1.value
    bx2 = x2.value
    minf0b = (-1)*bx1 - bx2    
    np.set_printoptions(precision=6, suppress=False)
    print("Best solution x1* = ", bx1)
    print("Best solution x2* = ", bx2)
    print("Best objective f0* = ",minf0b)

```


(c) $f_0(x_1, x_2) = x_1$.


```python
f0c = x1

cost = cp.Minimize(f0c)

prob = cp.Problem(cost, const)
prob.solve()

print("Solver Status:", prob.status)
if prob.status != 'optimal':
    print("⚠️ Optimization did not converge. Results are invalid.")
else:
    cx1 = 0
    if x1.value > 1e-6 :
        cx1 = x1.value
    cx2 = x2.value
    minf0c = cx1  
    np.set_printoptions(precision=6, suppress=False)
    print("Best solution x1* = ", cx1)
    print("Best solution x2* = ", cx2)
    print("Best objective f0* = ",minf0c)

```


(d) $f_0(x_1, x_2) = \max\{x_1, x_2\}$.


```python
f0d = cp.maximum(x1,x2)

cost = cp.Minimize(f0d)

prob = cp.Problem(cost, const)
prob.solve()

print("Solver Status:", prob.status)
if prob.status != 'optimal':
    print("⚠️ Optimization did not converge. Results are invalid.")
else:
    dx1 = x1.value
    dx2 = x2.value
    minf0d = np.maximum(dx1,dx2)    
    np.set_printoptions(precision=6, suppress=False)
    print("Best solution x1* = ", dx1)
    print("Best solution x2* = ", dx2)
    print("Best objective f0* = ",minf0d)

```


(e) $f_0(x_1, x_2) = x_1^2 + 9x_2^2$.

```python
f0e = (x1**2)+(9*(x2**2))

cost = cp.Minimize(f0e)

prob = cp.Problem(cost, const)
prob.solve()

print("Solver Status:", prob.status)
if prob.status != 'optimal':
    print("⚠️ Optimization did not converge. Results are invalid.")
else:
    ex1 = x1.value
    ex2 = x2.value
    minf0e = (ex1**2)+(9*(ex2**2))
    np.set_printoptions(precision=6, suppress=False)
    print("Best solution x1* = ", ex1)
    print("Best solution x2* = ", ex2)
    print("Best objective f0* = ",minf0e)

```
